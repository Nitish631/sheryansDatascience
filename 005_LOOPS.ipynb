# `range()` function:
- Return the sequence of the items
- syntax
```python
    range(start,stop,step)
```
- The stop point is exclusive
- The start point is inclusive

# `for` loop:
- It is used to iterate the sequencial datatype
- Usually use range function for the seuential data
- Use it if you know the no of steps have to use.

# `while` loop:
- It is used to perform a looop until the condition are satisfied.
- Needed a condition and a terminating operation
- syntax
```python
    while(condition):
        terminating operation
```

# `continue` keyword:
- It is used to escape an iteration
# `break` keyword:
- It is used to break the whole loop. the entire loop will be stop from that position
# `else` keyword:
- If the `break` keyword is encountered `else` block will not execute.
- But if no `break` is encountered `else` block will definitely execute

````python
def longest_consecutive(*nums):
    s=sorted(nums)
    r=0
    l=0
    count=0
    for i in range(len(s)-1):
        if s[i]==s[i+1]:continue
        if s[i]+1==s[i+1]:
            r+=1
        else:
            l=i+1
            r=i+1
        count=max(r-l+1,count)
    return count
print(longest_consecutive(100,4,200,1,3,2))
longest_consecutive(1,3,2,0,1)
````

````python
def move_zeros_to_end(nums):
    lst_temp=list(nums)
    for i in range(len(lst_temp)-1):
        for j in range(i,len(lst_temp)-1):
            if lst_temp[j]==0:
                temp=lst_temp[j+1]
                lst_temp[j+1]=lst_temp[j]
                lst_temp[j]=temp
    return "".join(map(str,lst_temp))
move_zeros_to_end()
````
