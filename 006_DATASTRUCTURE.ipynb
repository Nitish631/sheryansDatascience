### Reference vs shalllow vs deep copy

- Reference copy
```python
a ──┐
    ├──> [1,2,4,6,8]
b ──┘

a ──┐
    ├──> [[10,20],[30,40]]
b ──┘
```

````python
a=[1,2,4,6,8]
b=a
b[0]=100
print(a)
print(b)
print()

a=[[10,20],[30,40]]
b=a
b[1][0]=44
print(a)
print(b)
````

- Shallow copy
```python
a ───> [10,20,30,50]
b ───> [10,20,30,50]

a ──> [A , B]
b ──> [A , B]
```

````python
a=[10,20,30,50]
b=a.copy()
b[0]=100
print(a)
print(b)
print()
## ANOTHER EXAMPLE 
a=[[10,20],[30,40]]
b=a.copy()
b[1][0]=44
print(a)
print(b)
````

- Deep copy
```python
a ──> [A , B]
        │   │
        ▼   ▼
     [10,20] [30,40]

b ──> [C , D]
        │   │
        ▼   ▼
     [10,20] [30,40]
```

````python
import copy
a=[10,30,60,70]
b=copy.deepcopy(a)  
b[0]=100
print(a)
print(b)
print()
## ANOTHER EXAMPLE 
a=[[10,20],[30,40]]
b=copy.deepcopy(a)
b[1][0]=44
print(a)
print(b)
````

#### Quick Summary:
| Operation          | Outer List Copied? | Inner Lists Copied? |
| ------------------ | ------------------ | ------------------- |
| `b = a`            | ❌ No               | ❌ No                |
| `a.copy()`         | ✅ Yes              | ❌ No                |
| `copy.deepcopy(a)` | ✅ Yes              | ✅ Yes               |


### List traversing

````python
a=[10,30,50,70,39]
for i in a:
    print(i,end=",")
````

````python
for i in range(len(a)):
    print(a[i],end=",")
````
