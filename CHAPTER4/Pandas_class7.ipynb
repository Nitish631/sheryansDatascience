# 🐼 Pandas - Class 7: Merging, Joining & Concatenation
Welcome to **Class 7** of our Pandas series. Today we’ll learn how to combine multiple DataFrames using different techniques.

```python
import pandas as pd
from rich import print

df1 = pd.DataFrame({
    "Employee_ID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    "Name": ["Ram", "Sita", "Hari", "Ramesh", "Amit",
             "Nisha", "Bikash", "Puja", "Kiran", "Suman"],
    "Department": ["IT", "HR", "Finance", "IT", "Sales",
                    "HR", "IT", "Marketing", "Finance", "Sales"],
    "Salary": [60000, 45000, 55000, 70000, 48000,
               47000, 75000, 52000, 58000, 50000]
})
print(df1)

df2 = pd.DataFrame({
    "Employee_ID": [105, 106, 107, 108, 109, 110, 111, 112, 113, 114],
    "Age": [25, 29, 31, 24, 35, 28, 30, 26, 33, 27],
    "City": ["Kathmandu", "Pokhara", "Lalitpur", "Bhaktapur", "Kathmandu",
             "Pokhara", "Biratnagar", "Lalitpur", "Kathmandu", "Pokhara"],
    "Experience": [2, 4, 6, 1, 8, 3, 5, 2, 7, 4]
})
print(df2)

df3 = pd.DataFrame({
    "Score": [85, 78, 92, 88, 76, 90, 84, 95, 72, 89],
    "Rating": [4.2, 3.8, 4.7, 4.5, 3.6, 4.6, 4.1, 4.9, 3.5, 4.4]
})

print(df3)
```

## 1. Using `concat()` for Stacking Data
- `pd.concat()` combines DataFrames vertically (rows) or horizontally (columns).
- Use `axis=0` for vertical stacking (default), `axis=1` for horizontal.
- `ignore_index=True` can reset the index after concatenation.

```python

# Dataset 1: Product details


# Dataset 2: More products (same columns as df1 → good for vertical stacking)

# Dataset 3: Additional info (different columns → good for horizontal stacking)
```

```python
pd.concat([df1,df2])
```

```python
pd.concat([df1,df2],ignore_index=True)
```

```python
pd.concat([df1,df3])
```

```python
pd.concat([df1,df3],axis=1)
```

## 2. Using `merge()` for SQL-Style Joins
- `merge()` is similar to SQL joins.
- Specify `on` or `left_on`/`right_on` to choose key columns.
- Join types: `inner` (default), `left`, `right`, `outer`.

```python
pd.merge(df1,df2,on="Employee_ID",how="inner")
```

```python
pd.merge(df1,df2,on="Employee_ID",how="outer")
```

```python
pd.merge(df1,df2,on="Employee_ID",how="left")
```

```python
pd.merge(df1,df2,on="Employee_ID",how="right")
```

## 3. Using `join()` with Index Alignment
- `join()` joins DataFrames on their index or on a key column.
- By default, joins on index; can specify `on='col'`.
- Useful for adding columns to an existing DataFrame.

```python
emp=df1.set_index("Employee_ID")
df3_temp=df3.set_index(emp.index)
```

```python
emp.join(df3)
```

## 4. Practical Examples
- Combine sales and revenue data to see performance.
- Merge students with marks to link records.
- Practice with realistic datasets to understand alignment and missing data handling.

```python
sales = pd.DataFrame({
    "ProductID": [111, 102, 113, 104, 105, 106, 107, 108, 109, 110],
    "UnitsSold": [25, 40, 15, 30, 50, 20, 35, 45, 10, 28]
})

print(sales)

revenue = pd.DataFrame({
    "ProductID": [101, 102, 103, 104, 105, 116, 107, 108, 119, 110],
    "Revenue": [25000, 48000, 18000, 36000, 62500, 22000, 42000, 54000, 12000, 35000]
})

print(revenue)
```

```python
merged=pd.merge(sales,revenue,on="ProductID",how="outer")
print(merged)
```
