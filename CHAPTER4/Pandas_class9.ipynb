## 1. Applying Custom Functions with `apply()` / `map()`
- `apply()` lets you apply a custom function across rows or columns.
- `map()` is for applying functions element-wise to a Series.
- Both are handy for transformations, calculations, and cleaning data.

```python
import pandas as pd

data = {
    "Name": [
        "Ram", "Sita", "Hari", "Anita", "Bikash",
        "Nisha", "Aayush", "Puja", "Rohan", "Sneha",
        "Kiran", "Amit", "Prabin", "Suman", "Riya",
        "Nabin", "Sarita", "Roshan", "Manish", "Asha"
    ],
    "Math": [
        78, 85, 67, 92, 74,
        88, 95, 81, 69, 90,
        76, 84, 73, 89, 91,
        65, 87, 79, 94, 82
    ],
    "Science": [
        82, 79, 71, 95, 80,
        91, 89, 85, 74, 88,
        81, 86, 77, 92, 94,
        70, 83, 76, 90, 87
    ]
}

df = pd.DataFrame(data)

print(df)
```

```python
df["TotalMarks"]=df[["Math","Science"]].apply(lambda row:row.sum(),axis=1)
df
```

```python
col_mean=df[["Math","Science"]].apply(lambda col:col.mean())
col_mean
```

```python
def to_grade(s):
    if s>90:
        return "A"
    elif s>80:
        return "B"
    elif s>50:
        return "OK"
    else:
        return "Fail"
```

```python
df["Math_Grade"]=df["Math"].apply(to_grade)
df
```

```python
df["Name"]=df["Name"].map(str.lower)
df
```

```python
df["BonusMath"]=df["Math"].map(lambda x:x+5)
df
```
