##Fancy Indexing
you can Pick elements from specific positions using a list/array of indices.

```python
import numpy as np
arr=np.array([10,20,30,40,50,60,70,80])
a=[1,2,3,4,5,6,7,8,9]
print(a[0])
print(arr[[0,2,7]])
```

You can do this fancy indexing in 2D Matrix aswell

```python
matrix=np.arange(1,13).reshape(3,4)
print(matrix)
matrix[[0,2],[1,3]]# 1st element of 0th and 3rd element of 2nd array
```

## Conditional Selection
Find values based on conditions or replace them.

```python
arr=np.random.randint(2,33,8)
arr
```

```python
arr[arr<30]
```

```python
np.where(arr<30,arr,0)
```

## Sorting & Ordering
np.sort() → returns sorted array

np.argsort() → returns indices of sorted order

```python
np.sort(arr)
```

```python
np.argsort(arr)
```

## Concatenation & Stacking

```python
a=np.array([1,2,3])
b=np.array([4,5,6])
np.concat([a,b])
```

Vertical & Horizontal Stacking

```python
A=np.array([[1,2],[3,4]])
B=np.array([[5,6],[7,8]])
print(A)
print()
print(B)
```

- VERTICAL STACKING

```python
np.vstack((A,B))
```

```python
np.vstack((B,A))
```

- HORIZONTAL STACKING

```python
np.hstack((A,B))
```

```python
np.hstack((B,A))
```

```python
np.concat([A,B])
```
