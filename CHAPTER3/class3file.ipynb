##Numpy array operations

Lets first see how arrays behave like “vectors/matrices” when operations happen element-wise.

```python
import numpy as np
a=[1,2,3,4,5]
b=[5,6,7,8,9]
a+b
```

```python
a=np.array(a)
b=np.array(b)
print(a+b)
print(b-b)
print(a*b)
print(a/b)
print(b//a)
```

These things are already different from List and if we talk about one more special power it is **Broadcasting**

```python
a=np.array([1,2,3,4,5])
print(a*2)
```

Even these things work with different shaped array the calculations will happen on every row. but dimensions must match

```python
arr=np.random.randint(3,20,(3,4))
print(arr)
print(arr*4)
```

#Aggregate Functions

These functions are used to calculate the statistics on arrays.

```python
arr=np.random.randint(3,40,10)
arr
```

```python
print(np.mean(arr))
print(np.sum(arr))
print(np.median(arr))
print(np.min(arr))
print(np.max(arr))
print(np.std(arr))
print(np.var(arr))
```

You can also do the axis based operations row-Wise and Colum-wise

```python
data=np.random.randint(1,4,(5,6))
data
```

```python
print(np.sum(data,axis=0))#COLUMN WISE
print(np.mean(data,axis=0))
print(np.median(data,axis=0))
print(np.sum(data,axis=1))#ROW
print(np.mean(data,axis=1))
print(np.median(data,axis=1))
```
