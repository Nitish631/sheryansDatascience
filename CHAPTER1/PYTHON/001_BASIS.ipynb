# Comments, Variable and Data Types:

### Single line comment: 
- It is done by using `# comment description`
### Multiple line comment:
- It is done by using `""" comment description """`

```python
# this is the single line comment
"""
This is the 
Multiple line comment
"""
```

# Variables:

```python
a=5
b=55
a+b
```

### Naming Conventions:
- camelCase
- PascalCase
- snake_case

### Rule for naming a variable:
- Cannot start with number
- Must not contain space
- only `_` is allowed among the special character
