# File Handling:

### PROJECT OS:

```python
from pathlib import Path
import shutil
import os

def create_folder():
    try:
        name=input("Please tell folder name:- ")
        p=Path(name)
        p.mkdir()
    except Exception as e:
        print(f"Sorry an error occured as {e}")
def read_file_folder():
    p=Path("")
    items=list(p.rglob("*"))
    for i,item in enumerate(items):
        print(f"{i+1} : {item}")
def update_folder():
    try:
        read_file_folder()
        old_name=input("Please tell me which folder you want to update:- ")
        p=Path(old_name)
        if p.exists() and p.is_dir():
            new_name=input("Please tell your new folder name:- ")
            new_p=Path(new_name)
            p.rename(new_p)
            print("Your folder name is updated successfully")
        else:
            print("Sorry no such folder exist")
    except Exception as err:
        print(err)
def delete_folder():
    try:
        read_file_folder()
        name=input("Please tell me which folder you want to delete")
        deleting_path=Path(name)
        if deleting_path.exists() and deleting_path.is_dir():
            # deleting_path.rmdir()
            shutil.rmtree(deleting_path)
            print("Folder successfully deleted")
        else:
            print("No such Folder Exist.")
    except Exception as err:
        print(err)
def create_file():
    try:
        read_file_folder()
        name=input("Enter the file name")
        p=Path(name)
        if not p.exists():
            with open(p,"w")as fs:
                data=input("Write the data you want to write in file:- ")
                fs.write(data)
            print("File created successfully.")
        else:
            print("Sorry this name file already exist")
    except Exception as err:
        print(err)

def read_file():
    read_file_folder()
    name=input("Enter the file you want to read.")
    p=Path(name)
    if p.exists():
        with open(name,"r") as fs:
            content=fs.read()
            print("Your file content is :-\n")
            print(content)
    else:
        print("File name you provided doesnot exist")
def update_file():
    read_file_folder()
    name=input("Enter file name: ")
    p=Path(name)
    if p.exists() and p.is_file():
        with open(name,'a+') as f:
            content=input("Enter the extra content: ")
            f.write(content)
            f.seek(0)
            data=f.read()
            print("UPDATED DATA")
            print(data)
def delete_file():
    read_file_folder()
    name=input("Tell your file name with extension: ")
    path=Path(name)
    if path.exists() and path.is_file():
        path.unlink()
        print("File deleted successfully")
    else:
        print("File doesnot exists.")
```

```python
print("OPTIONS:","--"*8)
print("""
1. Create a folder
2. Read file and folders
3. Update the folder
4. Delete the folder
5. Create a file
6. Read a file
7. Update a file
8. Delete a file
""")
choice=int(input("Please choose your option."))
if choice==1:
    create_folder()
if choice==2:
    read_file_folder()
if choice==3:
    update_folder()
if choice==4:
    delete_folder()
if choice==5:
    create_file()
if choice==6:
    read_file()
if choice==7:
    update_file()
if choice==8:
    delete_file()
```
