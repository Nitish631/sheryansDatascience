#### Linear Search

```python
a=[1,7,4,3,8,5,4,9,4,3,7,6]
target=55
for i in range(len(a)):
    if a[i]==target:
        print(i)
        break
else:
    print("Sorry no such element exist.")
```

### Binary search
- Only for the sorted List

```python
b=[2,5,7,9,22,55,77,99,333,555,777,999,2222,4444,6666,7777,9999]
target=777
left_i=0
right_i=len(b)-1
for i in range(len(b)):
    t=left_i+(right_i-left_i)//2
    print(left_i," , ",right_i)
    if target==b[t]:
        print(f"Index = {t}")
        break
    elif target<b[t]:
        right_i=t-1
    else:
        left_i=t+1
else:
    print("Sorry no such element found")
```

## Sorting:
- Buble sort
- Selection sort
- insersion sort
- merge sort
- Quick sort
- Heap sort

#### Buble sort:
- swap the greatest value iteration by iteration to the right side
- O(n2)

```python
c=b.copy()
print(c)
for j in range(len(c)-1):
    for i in range(len(c)-1-j):
        if c[i]>c[i+1]:
            c[i],c[i+1]= c[i+1],c[i]
print(c)
```

### Selection sort:
- find the index of the least value and then swap with first unsorted element 
- O(n2)

```python
c=b.copy()
min_ind=0
for i in range(len(c)-1):
    min_ind=i
    for j in range(i+1,len(c)):
        if c[j]<c[min_ind]:
            min_ind=j
    c[i],c[min_ind]=c[min_ind],c[i]
print(c)
```
