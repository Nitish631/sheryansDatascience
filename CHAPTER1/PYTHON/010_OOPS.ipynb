### INSTANCE METHOD
- IF THE METHOD IS ACCEPTING THE LOCATION OF THE INSTANCE (OR OBJECT),
- THEN THE METHOD IS KNOWN AS THE INSTANCE METHOD

````python
class Animal:
    gender="Male" 
    def __init__(self,name,age):
        self.name=name 
        self.age=age 
    name="lion"
    def info(self): ## INSTANCE METHOD
        print("this is a method")
obj=Animal(name="lion",age=20)
obj.info()
````

### CLASS METHOD
- It uses an decorator `@classmethod`
- It do not rely on instance specific data.

````python
class Animal:
    gender="Male" 
    def __init__(self,name,age):
        self.name=name 
        self.age=age 
    def info(self): ## INSTANCE METHOD
        print("this is a instance method")
    @classmethod
    def clmethod(cls): # CLASS METHOD
        print(f"{cls.gender} is your gender.")
        # print(f"{cls.name}") ## This is invalid because the `name` belongs to instance not to the class
Animal.clmethod()
obj=Animal(name="lion",age=20)
obj.info()
obj.clmethod()
````

- This is invalid because the `name` belongs to instance not to the class
- ```python 
    print(f"{cls.name})
```


- USE CLASS METHOD AS AN ALTERNATIVE CONSTRUCTOR

````python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    @classmethod
    def from_string(cls, data):
        name, age = data.split(",")
        return cls(name, int(age))
s = Student.from_string("Nitish,20")
print(s.name)
print(s.age)
````

### STATIC METHOD:
- It doesnot depend on any specific instance of the class 
- Unlike instance methods , which use self
- It is defined using the decorator `@staticmethod`.
- It doesnot take the self or cls parameter as its first argument.

````python
class Animal:
    gender="Male" 
    def __init__(self,name,age):
        self.name=name 
        self.age=age 
    def info(self): ## INSTANCE METHOD
        print("this is a instance method")
    @classmethod
    def clmethod(cls): # CLASS METHOD
        print(f"{cls.gender} is your gender.")
    @staticmethod
    def stmethod():#STATIC METHOD
        print("Hello this is the static method")
Animal.clmethod()
obj=Animal(name="lion",age=20)
obj.info()
obj.clmethod()
obj.stmethod()
Animal.stmethod()
````

# ENCAPSULATIONN IN OOPs:

### Public accessible:
- Accessible ever where.
- outside the class and subclass

````python
class Animal:
    name="lion" #PUBLICLY ACCESSIBLE
    def speak(self):
        print("hello i will roar")
obj=Animal()
print(obj.name)
obj.speak()
````

### Protected Accessible
- Accessible within the class and subclass 
- Not accessible outside the class
- The protected access modifier doesnot work strictly like other programming languages like java or C++

````python
class Animal:
    _name="lion" #PROTECTED ACCESSIBLE
    def speak(self):
        print(f"Name = {self._name}") ## Accessible with in the class.
        print("hello i will roar")
obj=Animal()
obj.speak()
print(obj._name)#Still accessible with in the class. Not strictly restricted.
````

### Private Access modifier:
- No one can access the attribute or method made with this access modifier.
- It is only accessible within the class.

````python
class Animal:
    __name="lion" ## PRIVATE ACCESS MODIFIER
    def speak(self):
        print(f"Name = {self.__name}") ## Accessible with in the class.
        print("hello i will roar")
obj=Animal()
obj.speak()
#THE CONCEPT OF NAME MANGLING
a=obj._Animal__name
print(a)
#NOT DIRECTLY ACCESSIBLE
print(obj.__name)# Not accessible, strictly restricted.
````

# POLYMORPHISM:
- Same name having different form

````python
class Animal:
    name="lion"
    def speak(self):
        print("Lion roar")
class Bird:
    name="sparrow"
    def speak(self):
        print("bird tweet")
obj=Animal()
obj2=Bird()
obj.speak()
obj2.speak()
````

## Types of polymorphism:
- Method Overloading:
    - Python simulate method overloading using default or variable-length arguments, as it doesn't support traditional overloading.
- Method Overriding:
    - Occurs when a subclass defines a method with the same name as its superclass, replacing the superclass's method.

````python
class Animal:
    name="Lion"
    def speak(self):
        print("Hello I roar.")
class Human(Animal):
    name="Nitish"
    def speak(self):
        super().speak()## The super class's method can be called in the clild class
        print("Hello I speak.")
obj=Animal()
ob=Human()
obj.speak()
ob.speak()
````

# ABSTRACTION:
- Concept is to hide the unnecessary the internal implementation

### Decorator concept:

````python
import time
def timer(func):
    def wrapper(*args,**kwargs):
        start=time.time()
        result=func(*args,**kwargs)
        end=time.time()
        print(f"Function execution time is {end-start} sec.")
        return result
    return wrapper
@timer
def sumN(n):
    sum=0
    for i in range(1,n+1):
        sum+=i
    return sum

sumN(50)
````

- Abstraction is more about the Strictly implementation of the method in the child class rather than hiding the internal implementation of the method.

````python
from abc import ABC, abstractmethod
class Animal(ABC):
    @abstractmethod
    def sound(self):
        pass
class Dog(Animal):
    def sound(self):
        print("Hello I make woff sound")
class Cat(Animal):
    def sound(self):
        print("Hello I make meow sound.")
obj=Dog()
obj2=Cat()
obj.sound()
obj2.sound()
````

# DUNDER / MAGIC METHODS:
- Dunder methods are special methods in Python that define object behavior for built-in operations.
- They are prefixed with double underscores
- common dunder methods:
  - init
  - add
  - eq

### str method

````python
class Students:
    def __init__(self,name,marks):
        self.name=name
        self.marks=marks
    def __str__(self):
        return f"Name = {self.name}\nMarks = {self.marks}"
obj=Students("Nitish",33)
print(obj)
````

### len method

````python
class Shopping:
    def __init__(self,items):
        self.items=items
    def __len__(self):
        return len(self.items)
obj=Shopping(["apple","milk","bread"])
print(len(obj))
````

### add method

````python
class Numbers:
    def __init__(self,listNum):
        self.numbers=listNum
    def __add__(self, other):
        l1=len(self.numbers)
        l2=len(other.numbers)
        templist=[]
        if l1>l2:
            for i in range(l2):
                templist.append(self.numbers[i]+other.numbers[i])
            templist.extend(self.numbers[l2:])
        else:
            for i in range(l1):
                templist.append(self.numbers[i]+other.numbers[i])
            templist.extend(other.numbers[l1:])
        return templist
n1=Numbers(listNum=[1,2,3,4,5])
n2=Numbers(listNum=[1,2,3,4,5,6,7,8,9])
sum=n1+n2
print(sum)
````
