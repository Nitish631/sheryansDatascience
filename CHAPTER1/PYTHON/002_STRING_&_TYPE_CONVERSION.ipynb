# String and Type Conversion

## String

### String Indexing

```python
a="NATURE"
print(a[0])
print(a[-2])
```

### String slicing

```python
a="Nitish Rajbanshi"
print(a[0:6:1])
print(a[7::1])
```

- String is immutable.

### Print Statement ways:

```python
## STRING FORMATTING
age=20
designation="Data Scientist"
print(f"hello my age is {age} and my designation is {designation}.")
```

```python
### Escape sequence
print("Hello my name is Nitish.\nAnd my age is 23\tThe tab is used")
```

```python
### RAW STRING
print(r"Hello my name is Nitish.\nAnd my age is 23\tThe tab is used")
```

## Type Conversion

```python
a="23"
s=str(a)
f=float(a)
print(type(a))
print(type(s))
print(type(f))

a=23
b="good bye"
print(bool(a))
print(bool(0))
print(bool(-3))
print(bool(b))
print(bool(""))
```
