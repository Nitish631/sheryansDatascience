# OPERATORS

## Arithmetic Operators:
- `+ , - , * , / , // , % , **`

```python
print(5+44+55+22+5)
print(33-55)
print(5*6)
print(55/6)
print(55/5)
print(33//5)
print(40%3)
print(5**3)
```

## Assignment Operators / Shorthand Operators:
- `+= , -= , *= , /= , //= , %= `

```python
a=44
a+=2
print(a)
a-=2
print(a)
a*=2
print(a)
a/=2
print(a)
a//=5
print(a)
a%=5
print(a)
```

## Comparison Operators:
- ` == , != , > , >= , < , <= ,`

```python
print(12==12)
print(33!=33)
print(22>33)
print(22<33)
print(22>=22)
print(23>=22)
print(21>=22)
print(22<=22)
print(22==22!=12)
print(33>23>20)
print(55>33<30)
```

## Logical Operators:
- `and , or , not `

```python
print(3==3 and 5!=5)
print(3==3 or 5!=5)
print(not True)
```
