# Conditional Statement:

## `if-else` Statement

````python
age=int(input("Enter your age: "))
if age>=18:
    print("Eligible")
else:
    print("Not eligible")
````

## `pass` statement

````python
if age<18:
    pass
````

## Ternary operator in python:
- syntax
```python
    a = trueValue if condition else falseValue
```

````python
a = 55 if age>18 else 33
print(a)
````

````python
money=15
if money>20:
    print("Buy Cornato")
elif 10<=money<=20:
    print("Buy Bhujya")
else:
    print("Buy chocolate")
````
